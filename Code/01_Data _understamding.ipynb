# Module 8: Inventory & ABC Analysis

## Objective

Identify the most valuable products using ABC Analysis to improve inventory management and business decision-making.

---

## Business Questions

- Which products generate the highest revenue?
- Which products should receive inventory priority?
- Can we reduce inventory costs using ABC Analysis?

```python
product_sales = (
    df.groupby('Product Name')['Sales per customer']
    .sum()
    .sort_values(ascending=False)
)

product_sales.head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(14,6))

product_sales.head(10).plot(kind='bar')

plt.title("Top 10 Products by Sales")

plt.xlabel("Product")

plt.ylabel("Sales")

plt.xticks(rotation=90)

plt.show()
```

```python
abc = product_sales.reset_index()
abc.columns = ['Product', 'Sales']
```

```python
abc['Sales Percentage'] = (
    abc['Sales'] / abc['Sales'].sum()
) * 100
```

```python
abc['Cumulative Percentage'] = (
    abc['Sales Percentage'].cumsum()
)
```

```python
abc.head()
```

```python
def abc_category(x):

    if x <= 80:
        return "A"

    elif x <= 95:
        return "B"

    else:
        return "C"
```

```python
abc['Category'] = abc['Cumulative Percentage'].apply(abc_category)

abc.head()
```

```python
abc['Category'].value_counts()
```

```python
abc['Category'].value_counts().plot(kind='bar')

plt.title("ABC Inventory Classification")

plt.show()
```

# Observation

- A-category products contribute the majority of total sales.
- B-category products have moderate importance.
- C-category products contribute the least revenue.

# Business Insight

The company should prioritize inventory management for A-category products because they generate most of the revenue.

# Recommendation

- Maintain higher stock levels for A-category products.
- Monitor B-category products regularly.
- Reduce excess inventory for C-category products to minimize holding costs.
